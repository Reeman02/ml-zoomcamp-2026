## Q6 — Median value of horsepower

1. Calculate the original median.
2. Find the most frequent horsepower value.
3. Fill missing horsepower values with that mode.
4. Calculate the median again.


```python
median_before = df["horsepower"].median()
mode_horsepower = df["horsepower"].mode()[0]

df["horsepower"] = df["horsepower"].fillna(mode_horsepower)

median_after = df["horsepower"].median()

print("Median before:", median_before)
print("Mode:", mode_horsepower)
print("Median after:", median_after)
print("Changed:", median_before != median_after)
```

## Q7 — Sum of weights

This follows the homework instructions exactly:
- select cars from Asia
- select `vehicle_weight` and `model_year`
- take the first 7 rows
- convert to a NumPy array
- calculate `XTX`
- invert it
- multiply by `X.T` and `y`
- calculate the sum of `w`


```python
X = (
    df[df["origin"] == "Asia"]
    [["vehicle_weight", "model_year"]]
    .iloc[:7]
    .to_numpy()
)

XTX = X.T @ X

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = np.linalg.inv(XTX) @ X.T @ y

print("X:")
print(X)
print("\nXTX:")
print(XTX)
print("\nw:")
print(w)
print("\nSum of weights:", w.sum())
```
